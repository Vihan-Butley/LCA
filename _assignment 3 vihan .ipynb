```python
#ASSIGNMENT 3 -to check wether a figure is right angled tri 

def  right_angled_triangle():
  a = float(input("lenght of a: "))
  b = float(input("lenght of b: "))
  c = float(input("lenght of c: "))

  sides = [a,b,c]
  sides.sort()

#use pythagoras theorem
  if sides[0]**2 + sides[1]**2 == sides[2]**2 : 
     print("this is a right angle triangle")

  else :
     print("this is not a right angle triangle")

#call the function
right_angled_triangle()
 
```
